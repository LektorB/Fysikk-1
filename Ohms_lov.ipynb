# Ohms lov
Bruk en variabel motstand og still den inn på en tilfeldig verdi (ikke la denne verdien være svært liten). La motstanden ha den samme verdien gjennom hele forsøket. kobl motstanden til en variabel spenningskilde. Koble til et voltmeter og et ampermeter. Varier spenningen og noter ned ca. åtte sammenhørende strøm og spenningsverdier. Bruk programmet under til å plotte resultatet og beregne resistans med usikkerhet.


```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit 

#legger inn datasettet
spenning=[0]
strøm=[0]

#plotter datasettet som kryss
plt.plot(strøm,spenning,"xb")

#definer funksjonstypen med 
#rett antall parametre
def g(x,a):
    return a*x

#kurvetilpassing 
L,V=curve_fit(g, strøm, spenning)

print("her er parameterne i modellen: \n",L)
print("Standardavvik for parametrene over: \n",np.sqrt(np.diag(V)) )

#lager en array med x-verdier som går litt utenfor strøm,
#og tilsvarende funksjonsverdier ved hjelp av parmeterne i L og funksjonen 
deltax=max(strøm)-min(strøm)
x_verdi=np.linspace(0,max(strøm)+deltax*0.1,200)
y_verdi=g(x_verdi,*L)


plt.plot(x_verdi,y_verdi,"-b")
#plt.axhline(y=0,color="k") #akser dersom origo ikke er nede i venstre hjørne.
#plt.axvline(x=0,color="k")
plt.xlabel("I/$A$")
plt.ylabel("U/$V$")

# for proporsjonale sammenhenger gir det mening å plotte kurver med 
#proporsjonalitetskonstanten +/- to standatdavvik:

L_under=(L-np.sqrt(np.diag(V))*2)
y_under=g(x_verdi,*L_under)
plt.plot(x_verdi,y_under,"--r")

L_over=(L+np.sqrt(np.diag(V))*2)
y_over=g(x_verdi,*L_over)
plt.plot(x_verdi,y_over,"--r")

plt.grid(True,which='both')

plt.show()
```
