```python
# importerer to biblioteker og gir dem aliaser
import numpy as np 
import matplotlib.pyplot as plt

# Leser av data og lagrer det som en 2D array
data = np.loadtxt("ball2.txt",skiprows = 2,delimiter = ',',dtype='str',encoding = 'utf-8' )
data = np.char.replace(data, '−', '-').astype('float64')

tid = data[:,0]
pos = data[:,1]
plt.plot(tid,pos)
plt.title('Posisjonsgraf - Sprettball')
plt.xlabel('$t$/s')
plt.ylabel('$s$/m')
plt.show()
```

```python
n = len(tid)
vel = np.zeros(n-1)

for i in range(1,n-1):
    delta_s = pos[i+1] - pos[i-1]
    delta_t = tid[i+1] - tid[i-1]
    v = delta_s/delta_t
    vel[i] = v

plt.plot(tid[:-1],vel)
plt.plot([0,7.5],[0,0],'--')
plt.title('fartsgrafgraf - Sprettball')
plt.xlabel('$t$/s')
plt.ylabel('$v$/(m/s)')
plt.show()
```

```python
n = len(tid)
aks = np.zeros(n-2)

for i in range(1,n-2):
    delta_v = vel[i+1] - vel[i-1]
    delta_t = tid[i+1] - tid[i-1]
    a = delta_v/delta_t
    aks[i] = a

plt.plot(tid[:-2],aks)
#plt.plot([0,7.5],[0,0],'--')
plt.title('akselerasjonsgraf - Sprettball')
plt.xlabel('$t$/s')
plt.ylabel('$v$/(m/s)')
plt.show()
```

```python
plt.plot(tid[:-2],aks)
plt.plot([0,7.5],[-9.81,-9.81],'--')
plt.title('fartsgrafgraf - Sprettball')
plt.xlabel('$t$/s')
plt.ylabel('$v$/(m/s)')
plt.ylim(-15,10)
plt.show()
```
